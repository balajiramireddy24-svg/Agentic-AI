```python
marks =[35,45,55,65,75]
s=pd.Series(marks)
print(s)
print(type(s))
print(s.dtype)
print(s.ndim)
print(s.shape)
print(s.size)
print(type(s.dtype))
print(s.index)

```

```python
#Data Frame ---> 2 Dimensional Represnetation 
marks =[35,45,55,65,75]
q=pd.DataFrame(marks)
print(q)
w=pd.DataFrame(marks,index=['A','B','C','D','E'])
print(w)
e=pd.DataFrame(marks,columns=['Marks'])
print(e)
print(e.shape)
print(e.size)
print(e.ndim)
```

```python
details={'Product':['Mobile','Laptops','Earphones','Headsetes'],
         'Sales':[30000,50000,1000,2000],
         'Stock':[20,15,95,10]}
w=pd.DataFrame(details)
print(w)
print(w.shape)
print(w.size)
print(w.ndim)
print(type(w))
print(w.dtypes)
print(w.select_dtype['int'])
```

```python
#To Acess Columns 
print(w.columns)
```

```python
w['Product']
#Group of columns 
w[['Product','Sales']]
w[['Product','Sales']];
```

```python
w.index
```

```python
w.loc[0]
print(w.loc[0])   #locating index Values
print(w.loc[0]['Product'])
print(w.loc[0][['Product','Sales']])
print(w.loc[2][['Sales','Stock']])
print(w.loc[3][['Product','Sales','Stock']])
print(w.loc[1,'Stock'])
```

```python
#iloc()  --> Selecting by Positions
print(w.index)    #As its defaults intger index Both loc(),iloc()
print(w.iloc[0])   
print(w.iloc[[2,3]])
print(w.iloc[[2,3]][['Product','Stock']])
```

```python
print(w.info())   #It Gives Complete Infromation about DataFrame 
#Summary of the Data Frame    --->Descriptive Statistical info
print(w.describe())
print(w.describe(include='object'))    # for object Type  --> Count, Unique 
#top  ->Most Repecated Objects 
```

```python
#Change the Datatype of Columns
w['Sales']=w['Sales'].astype('float')
print(w['Sales'].info())
```
